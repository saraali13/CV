## 69. Valid convolution concept — WHAT / WHEN
Use true valid convolution when no padding is wanted and only positions where the kernel fully fits should be calculated.

```python
# Conceptual NumPy implementation for a 3x3 kernel:
H, W = gray.shape
k = kernel.shape[0]
valid = np.empty((H-k+1, W-k+1), dtype=np.float32)
for y in range(H-k+1):
    for x in range(W-k+1):
        valid[y, x] = np.sum(gray[y:y+k, x:x+k] * kernel)
```

```python
from collections import deque

def grow_region(gray, seed, tolerance):
    h, w = gray.shape
    sy, sx = seed
    ref = int(gray[sy, sx])
    q = deque([seed])
    seen = np.zeros((h, w), dtype=bool)
    mask = np.zeros((h, w), dtype=np.uint8)
    seen[sy, sx] = True

    while q:
        y, x = q.popleft()
        if abs(int(gray[y, x]) - ref) <= tolerance:
            mask[y, x] = 255
            for dy in (-1, 0, 1):
                for dx in (-1, 0, 1):
                    if dy == 0 and dx == 0:
                        continue
                    ny, nx = y + dy, x + dx
                    if 0 <= ny < h and 0 <= nx < w and not seen[ny, nx]:
                        seen[ny, nx] = True
                        q.append((ny, nx))
    return mask
```

## 97. Region-growing seed/tolerance — WHAT / WHEN
Use a good seed inside the desired object; smaller tolerance is stricter, while larger tolerance can cause leakage.

```python
seed = (342, 213)   # (y, x)
mask = grow_region(gray, seed, tolerance=15)
```

# ⚠️ COMMON EXAM MISTAKES

1. **OpenCV uses BGR**, Matplotlib normally displays RGB.
2. Pixel indexing is **`image[y, x]`**, not `image[x, y]`.
3. Grayscale image shape is `(H,W)`; color image is `(H,W,3)`.
4. Most Gaussian kernels are odd-sized: `3×3`, `5×5`, `7×7`.
5. `HoughLinesP()` returns line segments `(x1,y1,x2,y2)`.
6. `HoughCircles()` returns approximately `(x,y,r)`.
7. HOG → **shape/gradient orientation**; LBP → **texture**.
8. Sobel X and Y are gradient components; magnitude combines them.
9. Canny thresholds are low/high thresholds.
10. True **valid convolution** has no padding and produces a smaller output.
11. `cv2.filter2D()` normally keeps the output size; `BORDER_CONSTANT` is constant/zero border handling, **not true valid convolution**.
12. Perspective transform needs **4 corresponding point pairs**.
13. Homography is **3×3**; affine matrix is **2×3**.
14. K-Means needs pixels reshaped into rows such as `(-1,3)`.
15. SIFT: **keypoints = locations**, **descriptors = numerical descriptions**.
16. Homography with RANSAC needs at least **4 non-collinear point pairs**.
17. `param2` in Hough circles controls detection strictness.
18. Region growing depends strongly on the **seed** and **tolerance**.
19. Watershed needs meaningful markers; distance transform helps find object cores.
20. Always release videos with `cap.release()` and `out.release()`.

# ⭐ ONE-LINE VIVA ANSWERS

**Convolution:** Sliding a kernel over an image and combining local pixel values.

**Thresholding:** Separating pixels into groups using intensity conditions.

**Segmentation:** Dividing an image into meaningful regions or objects.

**Sobel:** Calculates image intensity gradients.

**Canny:** Produces a thin, structured edge map using multi-stage edge detection.

**HOG:** Describes shape using local gradient-orientation histograms.

**LBP:** Describes texture using binary comparisons of neighboring pixels.

**Watershed:** Separates touching objects using marker-based region growth.

**K-Means:** Groups pixels into `K` clusters based on feature similarity.

**Hough Transform:** Uses voting to detect parameterized geometric shapes.

**SIFT:** Detects local features and creates descriptors robust to scale and rotation.

**Homography:** A 3×3 projective mapping between two image planes.

**RANSAC:** Estimates a model while rejecting outlier correspondences.

**Wavelet:** Represents a signal at multiple scales/frequency components.

**Background subtraction:** Separates foreground changes from a learned background.
